# 1. Load the MNIST dataset

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
```

```python
(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()
```

# 2. Print shapes

```python
print("x_train shape:", x_train.shape)
print("y_train shape:", y_train.shape)
print("x_test shape:", x_test.shape)
print("y_test shape:", y_test.shape)
```

# 3. Reshape MNIST images to (28, 28, 1)

```python
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)
```

# 4. Normalize pixel values to range 0–1

```python
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
```

# 5. Create Sequential CNN model

```python
model = models.Sequential([
    layers.Conv2D(32, 3, activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

# 6. Compile the CNN

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

# 7. Train the model

```python
history = model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)
```

# 8. Evaluate the model on the test dataset

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
```

# 9. Print test accuracy and test loss

```python
print("Test Accuracy:", test_accuracy)
print("Test Loss:", test_loss)
```

# 10. Predict the class of the first image in the test dataset

```python
import numpy as np
```

```python
pred = model.predict(x_test[0].reshape(1, 28, 28, 1))
print("Predicted Class:", np.argmax(pred))
print("Actual Class:", y_test[0])
```
